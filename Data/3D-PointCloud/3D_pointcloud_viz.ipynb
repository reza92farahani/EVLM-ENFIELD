```python
import torch
from collections import Counter
```

## What the data includes:

```python
path = '/content/twoTowers_sample_0.pt'

o = torch.load(path, map_location="cpu", weights_only=False)

print("type string :", repr(o["type"]))

lab = o["semantic_labels"]
counts = Counter(lab.tolist())
tot = lab.numel()
print("\nsemantic label distribution:")
for c in sorted(counts):
    print(f"  class {c}: {counts[c]:7d} pts  ({100*counts[c]/tot:5.2f}%)")

print("\nobj_boxes (raw):")
print(o["obj_boxes"])
print("shape:", o["obj_boxes"].shape)
```

```python
import glob, os, torch
for f in sorted(glob.glob(path)):
    o = torch.load(f, map_location="cpu", weights_only=False)
    print(f"{os.path.basename(f):24s}  type={o['type']!r:16s}  "
          f"n_points={o['input_pcd'].shape[0]:6d}  n_boxes={o['obj_boxes'].shape[0]}")
```

## Visuzalization

```python
import plotly.graph_objects as go
o = torch.load(path, map_location="cpu", weights_only=False)
xyz = o["input_pcd"].float().numpy()
lab = o["semantic_labels"].numpy()
```

```python
import numpy as np

# Sample a smaller number of points for visualization
sample_size = 100000

# Get the total number of points
total_points = xyz.shape[0]

# Generate random indices for sampling
np.random.seed(42) # for reproducibility
sample_indices = np.random.choice(total_points, sample_size, replace=False)

# Create sampled xyz and labels
xyz_sampled = xyz[sample_indices]
lab_sampled = lab[sample_indices]

fig_sampled = go.Figure(go.Scatter3d(
    x=xyz_sampled[:,0], y=xyz_sampled[:,1], z=xyz_sampled[:,2], mode="markers",
    marker=dict(size=1.5, color=lab_sampled, colorscale="Turbo", showscale=True)))
fig_sampled.update_layout(scene_aspectmode="data", height=650, margin=dict(l=0,r=0,t=0,b=0))
fig_sampled.show()
```
